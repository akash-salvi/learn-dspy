## Signatures

Signatures are a way of expressing what you want an LLM to do by defining the desired input and outputs. They can be as simple as strings, like:

```
question -> answer
```

You can also specify your types as well, like:

```
sentence -> sentiment: bool
```

> *Note: By default type is always str*


Signatures can use whatever terms you’d like, so long as they’re descriptive of your desired inputs and outputs. For example:

```
document -> summary
novella -> tldr
baseball_player -> affiliated_team
```

Signatures can also be defined as a class, which lets you add further specs for more complex tasks.

````python
summarize = dspy.Predict('document -> summary')

document = """
The Renaissance, spanning roughly from the 14th to the 16th century, was a pivotal period in European history that marked the transition from the Middle Ages to modernity. Originating in Florence, Italy, its name literally means "rebirth," reflecting a renewed interest in the classical arts, literature, and philosophy of ancient Greece and Rome. This revival was fueled by the rediscovery of classic texts and a shift from theocentric (God-centered) to anthropocentric (human-centered) thought.

A key characteristic of this era was Humanism, a movement that emphasized human potential and achievement rather than divine or religious matters. This intellectual shift spurred incredible innovations across all fields. In art, masters like Leonardo da Vinci (known for The Last Supper and Mona Lisa) and Michelangelo (creator of the Sistine Chapel ceiling) perfected techniques such as perspective and realism. Their work moved away from the stylized, religious art of the past toward naturalistic representation.

Beyond the arts, the Renaissance fostered major advancements in science and politics. Nicolaus Copernicus challenged the long-held Ptolemaic model by proposing a heliocentric model of the universe, placing the sun, not the Earth, at the center. Furthermore, the invention of the printing press by Johannes Gutenberg around 1440 dramatically increased the spread of knowledge and literacy across the continent. This technological leap was instrumental in propagating Renaissance ideas and ultimately paving the way for the Reformation and the Age of Enlightenment.
"""

response = summarize(document=document)

print("Summary: ", response.summary)
````

### Multiple Inputs & outputs

````python
qna_with_context = dspy.Predict('question, context -> place, time')

question = "When and where was the Renaissance originated?"
context = """
The Renaissance, spanning roughly from the 14th to the 16th century, was a pivotal period in European history that marked the transition from the Middle Ages to modernity. Originating in Florence, Italy, its name literally means "rebirth," reflecting a renewed interest in the classical arts, literature, and philosophy of ancient Greece and Rome. This revival was fueled by the rediscovery of classic texts and a shift from theocentric (God-centered) to anthropocentric (human-centered) thought.

A key characteristic of this era was Humanism, a movement that emphasized human potential and achievement rather than divine or religious matters. This intellectual shift spurred incredible innovations across all fields. In art, masters like Leonardo da Vinci (known for The Last Supper and Mona Lisa) and Michelangelo (creator of the Sistine Chapel ceiling) perfected techniques such as perspective and realism. Their work moved away from the stylized, religious art of the past toward naturalistic representation.

Beyond the arts, the Renaissance fostered major advancements in science and politics. Nicolaus Copernicus challenged the long-held Ptolemaic model by proposing a heliocentric model of the universe, placing the sun, not the Earth, at the center. Furthermore, the invention of the printing press by Johannes Gutenberg around 1440 dramatically increased the spread of knowledge and literacy across the continent. This technological leap was instrumental in propagating Renaissance ideas and ultimately paving the way for the Reformation and the Age of Enlightenment.
"""

response = qna_with_context(question=question, context=context)

print("Place: ", response.place)
print("Time: ", response.time)
````

### Multiple Inputs & outputs with types

````python
emotions = dspy.Predict('input -> sentiment: str, confidence: float, reason: str, angry: bool, sad: bool')

response = emotions(input="I have been playing pubg for 6 hours but my KD is not crossing 1, its feels like to break my mobile")

print("Sentiment: ", response.sentiment)
print("Confidence: ", response.confidence)
print("Reason: ", response.reason)
print("Is he currently angry? ->", response.angry)
print("Is he currently sad? ->", response.sad)
````
