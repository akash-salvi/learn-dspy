```python
import os
import dspy
```

```python
API_KEY = os.getenv("GEMINI_API_KEY", "YOUR_GEMINI_API_KEY") 
```

```python
gemini = dspy.LM(model='gemini/gemini-2.5-flash', api_key=API_KEY)
        
# Configure DSPy to use this Language Model instance globally
dspy.configure(lm=gemini)
```

```python
gemini(messages=[{'role': 'user', 'content': 'why sky is blue'}])
```
